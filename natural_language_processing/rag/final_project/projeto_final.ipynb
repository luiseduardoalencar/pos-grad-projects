# 📦 PARTE 1: COLETA E PROCESSAMENTO DE DADOS

## 1.1 Montagem do Google Drive

**Objetivo:** Conectar ao Google Drive para persistência de dados

**Processo:** Montagem do drive e criação da estrutura de diretórios

````python
from bs4 import BeautifulSoup
import json
import re
````

````python
# ==========================================
# PROCESSAR ARQUIVOS HTML LOCAIS (CORRIGIDO)
# ==========================================

from bs4 import BeautifulSoup
import json
import re
import os

# Arquivos que você tem
arquivos = {
    'CDC': '/content/drive/MyDrive/pos-grad/neural_language_processing/data/cdc.htm',
    'CLT': '/content/drive/MyDrive/pos-grad/neural_language_processing/data/clt.htm',
    'CC': '/content/drive/MyDrive/pos-grad/neural_language_processing/data/cdf.html'
}

especialidades = {
    'CDC': 'Direito do Consumidor',
    'CLT': 'Direito Trabalhista',
    'CC': 'Direito Civil'
}

todos_documentos = []

for nome_lei, caminho in arquivos.items():
    print(f"📄 Processando {nome_lei}...")

    try:
        # Tentar diferentes encodings
        html = None
        for encoding in ['iso-8859-1', 'latin-1', 'cp1252', 'utf-8']:
            try:
                with open(caminho, 'r', encoding=encoding) as f:
                    html = f.read()
                print(f"  ✓ Encoding detectado: {encoding}")
                break
            except:
                continue

        if html is None:
            print(f"  ✗ Não foi possível ler o arquivo")
            continue

        # Processar HTML
        soup = BeautifulSoup(html, 'html.parser')
        texto = soup.get_text()

        # Dividir por artigos
        artigos = re.split(r'(Art\.?\s*\d+[°ºª]?\.?)', texto)

        # Processar pares
        for i in range(1, len(artigos)-1, 2):
            num_artigo = artigos[i].strip()
            conteudo = artigos[i+1].strip()

            if len(conteudo) > 50:
                todos_documentos.append({
                    'id': f'{nome_lei}_art_{len([d for d in todos_documentos if d["lei"] == nome_lei])+1}',
                    'lei': nome_lei,
                    'artigo': num_artigo,
                    'conteudo': conteudo[:1500],
                    'especialidade': especialidades[nome_lei],
                    'tamanho_original': len(conteudo)
                })

        qtd = len([d for d in todos_documentos if d['lei'] == nome_lei])
        print(f"✅ {nome_lei}: {qtd} artigos extraídos")

    except Exception as e:
        print(f"❌ Erro ao processar {nome_lei}: {e}")

print(f"\n🎉 TOTAL: {len(todos_documentos)} documentos extraídos!")

# Salvar
output_dir = '/content/drive/MyDrive/pos-grad/neural_language_processing/processed_data'
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, 'documentos_juridicos.json')

with open(output_path, 'w', encoding='utf-8') as f:
    json.dump(todos_documentos, f, ensure_ascii=False, indent=2)

print(f"💾 Salvo em: {output_path}")

print("\n📊 ESTATÍSTICAS:")
for lei in ['CDC', 'CLT', 'CC']:
    qtd = len([d for d in todos_documentos if d['lei'] == lei])
    print(f"  {lei}: {qtd} artigos")

if len(todos_documentos) > 0:
    print(f"\n📄 Exemplo: {todos_documentos[0]['artigo']}")
    print(f"Conteúdo: {todos_documentos[0]['conteudo'][:200]}...")
````

# 🤖 PARTE 4: SISTEMA RAG (RETRIEVAL-AUGMENTED GENERATION)

## 4.1 Geração de Embeddings com Legal-BERTimbau

**Objetivo:** Transformar documentos jurídicos em representações vetoriais para busca semântica

**Modelo Utilizado:** `rufimelo/Legal-BERTimbau-base`
- Modelo BERT especializado em textos jurídicos brasileiros
- Treinado em decisões judiciais, leis e acórdãos
- Dimensão dos embeddings: 768

**Processo:**
1. Carregamento do modelo Legal-BERTimbau
2. Preparação dos textos (artigo + conteúdo)
3. Geração de embeddings em lotes (batch_size=32)
4. Conversão para array NumPy

**Tempo estimado:** 5-10 minutos para 4.048 documentos

---

## 4.2 Criação do Banco Vetorial ChromaDB

**Objetivo:** Indexar embeddings para busca semântica eficiente

**Tecnologia:** ChromaDB
- Banco de dados vetorial de código aberto
- Métrica de similaridade: Cosseno
- Persistência em memória (projeto acadêmico)

**Processo:**
1. Inicialização do cliente ChromaDB
2. Criação da coleção "documentos_juridicos"
3. Inserção de documentos, embeddings e metadados
4. Indexação automática para busca rápida

---

## 4.3 Implementação da Busca por Palavra-Chave (BM25)

**Objetivo:** Complementar busca semântica com relevância lexical

**Algoritmo:** BM25 (Best Matching 25)
- Variação do TF-IDF
- Considera frequência de termos e comprimento do documento
- Eficaz para correspondências exatas de termos jurídicos

**Processo:**
1. Tokenização de todos os documentos
2. Criação do índice BM25
3. Cálculo de scores de relevância por palavra-chave

---

## 4.4 Busca Híbrida (Semântica + BM25)

**Objetivo:** Combinar melhor dos dois mundos: semântica e lexical

**Metodologia - RRF (Reciprocal Rank Fusion):**
- Busca semântica (ChromaDB): Captura similaridade de significado
- Busca por palavra-chave (BM25): Captura termos específicos
- Fusão por ranking recíproco: Score = peso_sem × (1/rank_sem) + peso_bm25 × (1/rank_bm25)
- Peso padrão: 70% semântica + 30% BM25

**Vantagens:**
- Maior robustez em diferentes tipos de consulta
- Combina compreensão contextual com precisão terminológica
- Ideal para domínio jurídico (termos técnicos + conceitos)

**Saída:** Top-K documentos mais relevantes com scores combinados

````python
import json
import numpy as np
from sentence_transformers import SentenceTransformer
import chromadb
from chromadb.config import Settings
from rank_bm25 import BM25Okapi
import re
````

````python
# ==========================================
# SISTEMA RAG COM LEGAL-BERTIMBAU
# ==========================================

# Instalar dependências

print("Carregando dados e modelos...")

# Carregar documentos
with open('/content/drive/MyDrive/pos-grad/neural_language_processing/processed_data/documentos_juridicos.json', 'r', encoding='utf-8') as f:
    documentos = json.load(f)

print(f"Total de documentos: {len(documentos)}")

# ==========================================
# 1. EMBEDDINGS COM LEGAL-BERTIMBAU
# ==========================================

print("\n1. Carregando Legal-BERTimbau...")

# Carregar modelo de embeddings especializado em direito brasileiro
embedding_model = SentenceTransformer('rufimelo/Legal-BERTimbau-base')

print("✅ Modelo carregado!")

# Gerar embeddings (pode demorar alguns minutos)
print("\n2. Gerando embeddings dos documentos...")
print("   (Isso pode levar alguns minutos...)")

# Preparar textos para embeddings
textos_para_embedding = []
for doc in documentos:
    # Combinar artigo + conteúdo para melhor contexto
    texto = f"{doc['artigo']}: {doc['conteudo']}"
    textos_para_embedding.append(texto)

# Gerar embeddings em lotes para otimizar
batch_size = 32
embeddings = []

for i in range(0, len(textos_para_embedding), batch_size):
    batch = textos_para_embedding[i:i+batch_size]
    batch_embeddings = embedding_model.encode(batch, show_progress_bar=True)
    embeddings.extend(batch_embeddings)

    if (i // batch_size) % 10 == 0:
        print(f"   Processados {i}/{len(textos_para_embedding)} documentos...")

embeddings = np.array(embeddings)
print(f"✅ Embeddings gerados: shape {embeddings.shape}")

# ==========================================
# 2. CHROMADB - VECTOR DATABASE
# ==========================================

print("\n3. Criando ChromaDB...")

# Criar cliente ChromaDB
chroma_client = chromadb.Client(Settings(
    anonymized_telemetry=False,
    allow_reset=True
))

# Deletar coleção se existir
try:
    chroma_client.delete_collection("documentos_juridicos")
except:
    pass

# Criar coleção
collection = chroma_client.create_collection(
    name="documentos_juridicos",
    metadata={"hnsw:space": "cosine"}
)

# Adicionar documentos ao ChromaDB
print("   Adicionando documentos ao ChromaDB...")

ids = [doc['id'] for doc in documentos]
documents_text = textos_para_embedding
metadatas = [
    {
        'lei': doc['lei'],
        'artigo': doc['artigo'],
        'especialidade': doc['especialidade']
    }
    for doc in documentos
]

collection.add(
    ids=ids,
    embeddings=embeddings.tolist(),
    documents=documents_text,
    metadatas=metadatas
)

print(f"✅ ChromaDB criado com {collection.count()} documentos")

# ==========================================
# 3. BM25 - BUSCA POR PALAVRA-CHAVE
# ==========================================

print("\n4. Criando índice BM25...")

# Tokenizar documentos para BM25
def tokenizar(texto):
    # Remover pontuação e converter para minúsculas
    texto = re.sub(r'[^\w\s]', ' ', texto.lower())
    # Dividir em palavras
    return texto.split()

textos_tokenizados = [tokenizar(texto) for texto in textos_para_embedding]
bm25 = BM25Okapi(textos_tokenizados)

print("✅ Índice BM25 criado")

# ==========================================
# 4. FUNÇÃO DE BUSCA HÍBRIDA
# ==========================================

def busca_hibrida(query, top_k=5, peso_semantico=0.7):
    """
    Busca híbrida combinando busca semântica (ChromaDB) e palavra-chave (BM25)

    Args:
        query: Pergunta do usuário
        top_k: Quantidade de documentos a retornar
        peso_semantico: Peso da busca semântica (0-1), BM25 = 1-peso_semantico

    Returns:
        Lista de documentos relevantes
    """

    # 1. BUSCA SEMÂNTICA (ChromaDB)
    query_embedding = embedding_model.encode([query])
    resultados_semanticos = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=top_k * 2  # Buscar mais para depois combinar
    )

    # 2. BUSCA POR PALAVRA-CHAVE (BM25)
    query_tokens = tokenizar(query)
    scores_bm25 = bm25.get_scores(query_tokens)

    # Pegar top indices do BM25
    top_indices_bm25 = np.argsort(scores_bm25)[::-1][:top_k * 2]

    # 3. FUSÃO RRF (Reciprocal Rank Fusion)
    scores_combinados = {}

    # Adicionar scores semânticos
    for idx, doc_id in enumerate(resultados_semanticos['ids'][0]):
        rank = idx + 1
        scores_combinados[doc_id] = peso_semantico * (1 / rank)

    # Adicionar scores BM25
    peso_bm25 = 1 - peso_semantico
    for idx, doc_idx in enumerate(top_indices_bm25):
        doc_id = documentos[doc_idx]['id']
        rank = idx + 1

        if doc_id in scores_combinados:
            scores_combinados[doc_id] += peso_bm25 * (1 / rank)
        else:
            scores_combinados[doc_id] = peso_bm25 * (1 / rank)

    # Ordenar por score combinado
    docs_ordenados = sorted(
        scores_combinados.items(),
        key=lambda x: x[1],
        reverse=True
    )[:top_k]

    # Recuperar documentos completos
    resultado_final = []
    for doc_id, score in docs_ordenados:
        # Encontrar documento original
        doc_original = next(d for d in documentos if d['id'] == doc_id)
        resultado_final.append({
            'documento': doc_original,
            'score': score
        })

    return resultado_final

# ==========================================
# 5. TESTAR BUSCA HÍBRIDA
# ==========================================

print("\n5. Testando busca híbrida...")

queries_teste = [
    "O que é consumidor segundo o CDC?",
    "Como funcionam as férias trabalhistas?",
    "O que é prescrição no direito civil?"
]

for query in queries_teste:
    print(f"\n📝 Query: {query}")
    resultados = busca_hibrida(query, top_k=3)

    for i, resultado in enumerate(resultados, 1):
        doc = resultado['documento']
        score = resultado['score']
        print(f"\n  {i}. {doc['lei']} - {doc['artigo']} (Score: {score:.4f})")
        print(f"     {doc['conteudo'][:150]}...")

# ==========================================
# 6. SALVAR SISTEMA RAG
# ==========================================

print("\n6. Salvando configuração do sistema RAG...")

config_rag = {
    'modelo_embedding': 'rufimelo/Legal-BERTimbau-base',
    'total_documentos': len(documentos),
    'dimensao_embeddings': embeddings.shape[1],
    'database': 'ChromaDB',
    'busca': 'Híbrida (Semântica + BM25)'
}

with open('/content/drive/MyDrive/pos-grad/neural_language_processing/processed_data/config_rag.json', 'w', encoding='utf-8') as f:
    json.dump(config_rag, f, ensure_ascii=False, indent=2)

print("✅ Sistema RAG configurado e testado com sucesso!")
print(f"\nConfiguração:")
print(f"  - Modelo: {config_rag['modelo_embedding']}")
print(f"  - Documentos: {config_rag['total_documentos']}")
print(f"  - Dimensão: {config_rag['dimensao_embeddings']}")
````

````python
# ==========================================
# RECARREGAR COMPONENTES SALVOS
# ==========================================

import json
import numpy as np
from sentence_transformers import SentenceTransformer
from setfit import SetFitModel
import chromadb
from chromadb.config import Settings
from rank_bm25 import BM25Okapi
import re

print("Carregando componentes salvos...\n")

# 1. CARREGAR DOCUMENTOS
print("1. Carregando documentos...")
with open('/content/drive/MyDrive/pos-grad/neural_language_processing/processed_data/documentos_juridicos.json', 'r', encoding='utf-8') as f:
    documentos = json.load(f)
print(f"   ✅ {len(documentos)} documentos carregados")

# 2. CARREGAR EMBEDDINGS PRÉ-CALCULADOS
print("\n2. Carregando embeddings...")
embeddings = np.load('/content/drive/MyDrive/pos-grad/neural_language_processing/processed_data/embeddings.npy')
print(f"   ✅ Embeddings carregados: {embeddings.shape}")

# 3. CARREGAR MODELO DE EMBEDDINGS
print("\n3. Carregando modelo Legal-BERTimbau...")
embedding_model = SentenceTransformer('rufimelo/Legal-BERTimbau-base')
print(f"   ✅ Modelo carregado")

# 4. RECRIAR CHROMADB
print("\n4. Recriando ChromaDB...")
chroma_client = chromadb.Client(Settings(anonymized_telemetry=False, allow_reset=True))

try:
    chroma_client.delete_collection("documentos_juridicos")
except:
    pass

collection = chroma_client.create_collection(
    name="documentos_juridicos",
    metadata={"hnsw:space": "cosine"}
)

# Preparar textos
textos_para_embedding = [f"{doc['artigo']}: {doc['conteudo']}" for doc in documentos]

# Adicionar ao ChromaDB
ids = [doc['id'] for doc in documentos]
metadatas = [{'lei': doc['lei'], 'artigo': doc['artigo'], 'especialidade': doc['especialidade']}
             for doc in documentos]

collection.add(
    ids=ids,
    embeddings=embeddings.tolist(),
    documents=textos_para_embedding,
    metadatas=metadatas
)
print(f"   ✅ ChromaDB criado com {collection.count()} documentos")

# 5. RECRIAR BM25
print("\n5. Recriando índice BM25...")
def tokenizar(texto):
    texto = re.sub(r'[^\w\s]', ' ', texto.lower())
    return texto.split()

textos_tokenizados = [tokenizar(texto) for texto in textos_para_embedding]
bm25 = BM25Okapi(textos_tokenizados)
print(f"   ✅ Índice BM25 criado")

# 6. RECRIAR FUNÇÃO DE BUSCA
def busca_hibrida(query, top_k=5, peso_semantico=0.7):
    query_embedding = embedding_model.encode([query])
    resultados_semanticos = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=top_k * 2
    )

    query_tokens = tokenizar(query)
    scores_bm25 = bm25.get_scores(query_tokens)
    top_indices_bm25 = np.argsort(scores_bm25)[::-1][:top_k * 2]

    scores_combinados = {}

    for idx, doc_id in enumerate(resultados_semanticos['ids'][0]):
        rank = idx + 1
        scores_combinados[doc_id] = peso_semantico * (1 / rank)

    peso_bm25 = 1 - peso_semantico
    for idx, doc_idx in enumerate(top_indices_bm25):
        doc_id = documentos[doc_idx]['id']
        rank = idx + 1
        if doc_id in scores_combinados:
            scores_combinados[doc_id] += peso_bm25 * (1 / rank)
        else:
            scores_combinados[doc_id] = peso_bm25 * (1 / rank)

    docs_ordenados = sorted(scores_combinados.items(), key=lambda x: x[1], reverse=True)[:top_k]

    resultado_final = []
    for doc_id, score in docs_ordenados:
        doc_original = next(d for d in documentos if d['id'] == doc_id)
        resultado_final.append({'documento': doc_original, 'score': score})

    return resultado_final

print(f"   ✅ Função de busca recriada")

# 7. CARREGAR CLASSIFICADOR
print("\n6. Carregando classificador SetFit...")
modelo_classificador = SetFitModel.from_pretrained(
    '/content/drive/MyDrive/pos-grad/neural_language_processing/processed_data/classificador_setfit'
)

def classificar_query(query):
    return modelo_classificador.predict([query])[0]

print(f"   ✅ Classificador carregado")

print("\n" + "="*60)
print("✅ TODOS OS COMPONENTES CARREGADOS COM SUCESSO!")
print("="*60)
````

## 7.2 Métrica 1: Precisão do Sistema RAG

**Objetivo:** Avaliar se a fonte correta está entre os top-5 documentos recuperados

**Metodologia:**
- Dataset: 50 pares pergunta-resposta com fonte conhecida
- Para cada pergunta, buscar top-5 documentos
- Verificar se a fonte de referência está entre os 5
- Calcular percentual de acertos

**Métrica:** Precisão = (Quantidade de acertos) / (Total de perguntas)

**Critério de avaliação:**
- ≥80% = 7 pontos
- 70-79% = 5 pontos
- 60-69% = 3 pontos
- <60% = 0 pontos

````python
# ==========================================
# MÉTRICA 1: PRECISÃO
# ==========================================

print("Calculando Precisão do sistema RAG...\n")

# Carregar dataset RAG
with open('/content/drive/MyDrive/pos-grad/neural_language_processing/processed_data/dataset_rag_50_pares.json', 'r', encoding='utf-8') as f:
    dataset_rag = json.load(f)

print(f"Dataset carregado: {len(dataset_rag)} pares\n")

# Avaliar
acertos = 0
total = len(dataset_rag)
resultados_detalhados = []

for i, par in enumerate(dataset_rag, 1):
    pergunta = par['pergunta']
    fonte_esperada = par['fonte']

    # Buscar top-5
    resultados = busca_hibrida(pergunta, top_k=5)

    # Verificar se fonte está nos top-5
    fontes_recuperadas = [f"{r['documento']['lei']} - {r['documento']['artigo']}"
                          for r in resultados]

    acertou = fonte_esperada in fontes_recuperadas

    if acertou:
        acertos += 1

    resultados_detalhados.append({
        'id': i,
        'pergunta': pergunta,
        'fonte_esperada': fonte_esperada,
        'fontes_recuperadas': fontes_recuperadas,
        'acertou': acertou
    })

    if i % 10 == 0:
        print(f"Processadas {i}/{total} perguntas...")

# Calcular métrica
precisao_5 = (acertos / total) * 100

print("\n" + "="*60)
print("RESULTADO - PRECISÃO")
print("="*60)
print(f"Acertos: {acertos}/{total}")
print(f"Precisão: {precisao_5:.2f}%")
print("="*60)

# Mostrar exemplos de erros
erros = [r for r in resultados_detalhados if not r['acertou']]
if erros:
    print(f"\nExemplos de ERROS ({len(erros)} no total):")
    for erro in erros[:3]:
        print(f"\n  Pergunta: {erro['pergunta']}")
        print(f"  Esperada: {erro['fonte_esperada']}")
        print(f"  Recuperadas: {', '.join(erro['fontes_recuperadas'][:3])}")

# Determinar pontuação
if precisao_5 >= 80:
    pontos = 7
elif precisao_5 >= 70:
    pontos = 5
elif precisao_5 >= 60:
    pontos = 3
else:
    pontos = 0

print(f"\n📊 PONTUAÇÃO: {pontos}/7 pontos")
````

## 7.3 Métrica 2: BLEU Score

**Objetivo:** Avaliar qualidade das respostas geradas comparando com respostas de referência

**Metodologia:**
- Gerar respostas para as 50 perguntas do dataset
- Comparar com respostas de referência usando BLEU
- BLEU mede similaridade entre textos (0 a 1)

**Critério de avaliação:**
- ≥0.6 = 6 pontos
- 0.4-0.59 = 4 pontos
- 0.2-0.39 = 2 pontos
- <0.2 = 0 pontos

````python
# ==========================================
# MÉTRICA 2: BLEU SCORE
# ==========================================

!pip install -q nltk

from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
import nltk
nltk.download('punkt', quiet=True)

print("Calculando BLEU Score...\n")

def gerar_resposta_rag(pergunta):
    """Gera resposta baseada em busca RAG"""
    docs = busca_hibrida(pergunta, top_k=3)

    resposta = ""
    for resultado in docs:
        doc = resultado['documento']
        resposta += f"{doc['conteudo'][:200]} "

    return resposta.strip()

# Calcular BLEU para cada par
scores_bleu = []
smoothing = SmoothingFunction().method1

for i, par in enumerate(dataset_rag, 1):
    pergunta = par['pergunta']
    referencia = par['resposta_referencia']

    # Gerar resposta
    resposta_gerada = gerar_resposta_rag(pergunta)

    # Tokenizar
    ref_tokens = referencia.lower().split()
    gen_tokens = resposta_gerada.lower().split()

    # Calcular BLEU
    score = sentence_bleu([ref_tokens], gen_tokens, smoothing_function=smoothing)
    scores_bleu.append(score)

    if i % 10 == 0:
        print(f"Processadas {i}/{len(dataset_rag)} perguntas...")

# Média
bleu_medio = sum(scores_bleu) / len(scores_bleu)

print("\n" + "="*60)
print("RESULTADO - BLEU SCORE")
print("="*60)
print(f"BLEU Médio: {bleu_medio:.4f}")
print(f"BLEU Mínimo: {min(scores_bleu):.4f}")
print(f"BLEU Máximo: {max(scores_bleu):.4f}")
print("="*60)

# Determinar pontuação
if bleu_medio >= 0.6:
    pontos = 6
elif bleu_medio >= 0.4:
    pontos = 4
elif bleu_medio >= 0.2:
    pontos = 2
else:
    pontos = 0

print(f"\n📊 PONTUAÇÃO: {pontos}/6 pontos")
````
