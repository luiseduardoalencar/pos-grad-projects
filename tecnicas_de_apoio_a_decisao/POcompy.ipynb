```python
from gurobipy import Model, GRB
```

```python
#criar o modelo

m = Model()
m.Params.OutputFlag = 0
```

```python
#variaveis

x1 = m.addVar()
x2 = m.addVar()
```

```python
#Função objetivo

m.setObjective(50*x1 + 30*x2, GRB.MAXIMIZE)
```

```python
m.addConstr(4*x1 + 3*x2 <= 240)
m.addConstr(3*x1 + 1*x2 <= 100)
```

```python
m.optimize()
```

```python
print("Lucro máximo : ", m.ObjVal)
print("Mesas = ", x1.X)
print("Cadeiras = ", x2.X)
```
